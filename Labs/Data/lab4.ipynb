1 Load the loans and look for gaps
Load library_loans.csv with parse_dates for both date columns. Print the shape and the
dtypes. Then count the missing values in every column, and show them as percent. Which
two columns have gaps, and how many?

Answer:
loan_date       3
return_date    36

```python
import pandas as pd

loans_raw = pd.read_csv('library_loans_clean.csv', parse_dates=['loan_date', 'return_date'])
loans = loans_raw.copy()
pp(loans.shape)
pp(loans.info())

pp(loans.isna().sum())
pp(loans.isna().mean() * 100)
```

2 Two kinds of missing
Show the loans where loan_date is missing. Then look at return_date: how many are
missing and the book is back (returned is True), and how many are missing and the book
is not back? Which of the two kinds is "missing for a reason", and which is "missing by
accident"? Create a column has_return_date with notna() and count it.

```python
loan_date = loans['loan_date']
pp(loans.loc[loan_date.isna(), ['loan_id', 'loan_date']])

return_date = loans['return_date']
pp(return_date.isna().sum()) # 36

missing_by_accident = (return_date.isna()) & (loans['returned'] == True)
pp(loans[missing_by_accident]['loan_id'].count()) # 6

missing_for_a_reason = (return_date.isna()) & (loans['returned'] == False)
pp(loans[missing_for_a_reason]['loan_id'].count()) # 30

loans['has_return_date'] = loans['return_date'].notna()
pp(loans['has_return_date'].value_counts()) # 564 / 36

pp(loans.shape)
```

3 Look at the members
Load library_members.csv, make a copy, and print the shape. Run info() and describe()
on birth_year. Something is wrong with the minimum and the maximum. Count how
many members have each strange value, and check for duplicates in the whole row and in
member_id. Write the list of problems as a comment.

```python
members_raw = pd.read_csv('library_members.csv')
members = members_raw.copy()
pp(members.shape)
pp(members.info())
pp(members['birth_year'].describe())

pp((members['birth_year'] == 1900).sum()) # 8
pp((members['birth_year'] == 2030).sum()) # 3
pp(members.duplicated().sum()) # 4
pp(members['member_id'].duplicated().sum()) # 4
```

Problems:

1. 12 branches are missing
2. 18 birth years are missing
3. 3 birth years are 1900, a placeholder
4. 3 birth years are 2030, which is impossible
5. 4 duplicated rows

4 Placeholders and impossible values
Nobody in this library is born in 1900, and nobody is born in the future. Use where and
between(1920, 2015) to turn those values into missing values. How many birth years are
missing now? Run describe() again and compare it with before.

```python
members['birth_year'] = members['birth_year'].where(members['birth_year'].between(1920, 2015))
pp(members['birth_year'].isna().sum()) # 29
pp(members['birth_year'].describe()) # 119
```
