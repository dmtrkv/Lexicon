1 Load the loans and look for gaps
Load library_loans.csv with parse_dates for both date columns. Print the shape and the
dtypes. Then count the missing values in every column, and show them as percent. Which
two columns have gaps, and how many?

Answer:
loan_date       3
return_date    36

```python
import pandas as pd

loans_raw = pd.read_csv('library_loans_clean.csv', parse_dates=['loan_date', 'return_date'])
loans = loans_raw.copy()
pp(loans.shape)
pp(loans.info())

pp(loans.isna().sum())
pp(loans.isna().mean() * 100)
```

2 Two kinds of missing
Show the loans where loan_date is missing. Then look at return_date: how many are
missing and the book is back (returned is True), and how many are missing and the book
is not back? Which of the two kinds is "missing for a reason", and which is "missing by
accident"? Create a column has_return_date with notna() and count it.

```python
loan_date = loans['loan_date']
pp(loans.loc[loan_date.isna(), ['loan_id', 'loan_date']])

return_date = loans['return_date']
pp(return_date.isna().sum()) # 36

mask = (return_date.isna()) & (loans['returned'] == True) # missing by accident
pp(loans[(mask)]['loan_id'].count()) # 6

mask = (return_date.isna()) & (loans['returned'] == False) # missing for a reason
pp(loans[(mask)]['loan_id'].count()) # 30

loans['has_return_date'] = loans['return_date'].notna()
pp(loans['has_return_date'].value_counts()) # 564 / 36

loans.shape
```
